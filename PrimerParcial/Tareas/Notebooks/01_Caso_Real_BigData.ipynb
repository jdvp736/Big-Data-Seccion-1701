# Propósito de la actividad

En esta actividad se analizará cómo una empresa como Netflix utiliza Big Data no solo para recomendar contenido, sino también para decidir qué series y películas producir.

El objetivo es comprender cómo el análisis de datos influye directamente en decisiones estratégicas dentro de la industria del entretenimiento digital.

---

# ¿Qué datos genera?

Netflix genera una gran variedad de datos relacionados con el consumo de contenido por parte de los usuarios.

Entre los datos producidos se encuentran:

- Series y películas más vistas  
- Tiempo promedio de visualización  
- Contenido abandonado  
- Preferencias de géneros  
- Actores o directores más populares  
- Horarios de consumo  
- Regiones con mayor audiencia  
- Interacciones dentro de la plataforma  

El volumen de datos es muy alto, ya que millones de usuarios consumen contenido diariamente.

La frecuencia de generación de datos es constante, generándose información en tiempo real con cada reproducción.

---

# ¿Quién los usa?

Los datos de Netflix son utilizados por diferentes actores dentro de la empresa.

## Empresa

Netflix utiliza los datos para:

- Decidir qué contenido producir  
- Invertir en series y películas  
- Identificar tendencias globales  
- Reducir riesgos en nuevas producciones  

## Analistas y científicos de datos

Procesan la información para identificar patrones de consumo y predecir el éxito de futuros contenidos.

## Equipos creativos

Utilizan los datos como referencia para desarrollar historias, personajes y estilos que conecten con la audiencia.

---

# ¿Qué problema resuelve?

El uso de Big Data permite a Netflix reducir la incertidumbre en la producción de contenido.

Gracias a los datos, la empresa puede:

- Crear contenido con mayor probabilidad de éxito  
- Evitar inversiones en proyectos poco atractivos  
- Entender mejor a su audiencia  
- Aumentar la retención de usuarios  
- Competir con otras plataformas de streaming  

Además, permite tomar decisiones basadas en datos reales y no únicamente en intuición.

---

# ¿Por qué no bastaría Excel?

Excel no sería suficiente para manejar los datos utilizados por Netflix en la toma de decisiones de producción.

Esto se debe a que:

- Se manejan millones de registros diarios  
- Los datos provienen de múltiples fuentes  
- Se requiere análisis en tiempo real  
- Se utilizan modelos predictivos avanzados  

Además, Netflix trabaja con datos complejos como:

- Videos  
- Metadatos de contenido  
- Comportamiento del usuario  
- Datos globales de audiencia  

Por esta razón, se necesitan herramientas de Big Data que permitan procesar grandes volúmenes de información de forma rápida, eficiente y escalable.
