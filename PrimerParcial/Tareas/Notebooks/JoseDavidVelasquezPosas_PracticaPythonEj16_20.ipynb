# Ejercicio 16: DataFrame de estudiantes

**Enunciado:** crear un DataFrame con información de estudiantes y realizar análisis básicos.

**Lógica:** se crea un DataFrame con pandas, luego se utilizan métodos como `head()`, `info()`, `mean()` y filtros para analizar los datos. También se ordenan los registros por promedio.

```python
import pandas as pd

data = {
    "nombre": ["Ana","Luis","Carlos","Marta","Sofía","Pedro","Elena","Juan","Diego","Laura"],
    "carrera": ["Sistemas","Derecho","Medicina","Sistemas","Arquitectura","Sistemas","Derecho","Medicina","Sistemas","Arquitectura"],
    "edad": [20,22,21,23,19,24,22,21,23,20],
    "promedio": [85,78,90,88,76,95,80,70,92,84],
    "ciudad": ["Tegus","SPS","Tegus","SPS","La Ceiba","Tegus","SPS","La Ceiba","Tegus","SPS"]
}

df = pd.DataFrame(data)

print(df.head())
print("\nInformación:")
print(df.info())

print("\nPromedio general:", df["promedio"].mean())

print("\nEstudiantes >= 80:")
print(df[df["promedio"] >= 80])

print("\nOrdenado por promedio:")
print(df.sort_values(by="promedio", ascending=False))
```

# Ejercicio 17: Análisis de ventas con pandas

**Enunciado:** analizar un conjunto de datos de ventas.

**Lógica:** se crea un DataFrame, se calcula una nueva columna multiplicando cantidad por precio, luego se agrupan y analizan los datos para obtener totales y el producto más vendido.

```python
ventas = pd.DataFrame({
    "vendedor": ["Ana","Luis","Ana","Carlos","Luis","Ana"],
    "producto": ["Laptop","Mouse","Teclado","Laptop","Mouse","Laptop"],
    "categoria": ["Tecnología","Accesorios","Accesorios","Tecnología","Accesorios","Tecnología"],
    "cantidad": [2,5,3,1,4,2],
    "precio": [1000,20,50,1000,20,1000]
})

ventas["TotalVenta"] = ventas["cantidad"] * ventas["precio"]

print("\nVentas:")
print(ventas)

print("\nTotal por vendedor:")
print(ventas.groupby("vendedor")["TotalVenta"].sum())

print("\nProducto más vendido:")
print(ventas.groupby("producto")["cantidad"].sum().idxmax())

print("\nCategoría con mayor ingreso:")
print(ventas.groupby("categoria")["TotalVenta"].sum().idxmax())

print("\nOrdenado:")
print(ventas.sort_values(by="TotalVenta", ascending=False))
```

# Ejercicio 18: Limpieza de datos con pandas

**Enunciado:** limpiar un DataFrame con datos incompletos.

**Lógica:** se identifican valores nulos, se cuentan, se reemplazan valores vacíos y se eliminan registros incompletos usando funciones de pandas.

```python
clientes = pd.DataFrame({
    "nombre": ["Ana","Luis","Carlos","Marta"],
    "edad": [20,None,25,30],
    "correo": ["a@mail.com",None,"c@mail.com",""],
    "ciudad": ["Tegus","",None,"SPS"],
    "telefono": ["1234","5678",None,"9101"]
})

print("Valores nulos:")
print(clientes.isnull())

print("\nCantidad de nulos:")
print(clientes.isnull().sum())

clientes["ciudad"] = clientes["ciudad"].replace("", "No especificada")

clientes = clientes[clientes["correo"].notnull() & (clientes["correo"] != "")]

print("\nDataFrame limpio:")
print(clientes)
```

# Ejercicio 19: Análisis de préstamos

**Enunciado:** analizar datos de solicitudes de préstamo.

**Lógica:** se crea un DataFrame, se calculan promedios, se filtran datos por estado y se agrupan para obtener estadísticas.

```python
prestamos = pd.DataFrame({
    "cliente": ["Ana","Luis","Carlos","Marta"],
    "monto": [60000,30000,80000,45000],
    "plazo": [12,24,18,12],
    "ingreso": [2000,1500,3000,1800],
    "estado": ["Aprobado","Pendiente","Aprobado","Rechazado"]
})

print("Promedio solicitado:", prestamos["monto"].mean())

print("\nSolicitudes por estado:")
print(prestamos["estado"].value_counts())

print("\nPréstamos aprobados:")
print(prestamos[prestamos["estado"] == "Aprobado"])

print("\nMontos > 50000:")
print(prestamos[prestamos["monto"] > 50000])

print("\nPromedio por estado:")
print(prestamos.groupby("estado")["monto"].mean())
```

# Ejercicio 20: Mini análisis de empleados

**Enunciado:** analizar información de empleados en una empresa.

**Lógica:** se crea un DataFrame y se utilizan funciones como `describe()`, `groupby()` y filtros. También se crea una nueva columna según la experiencia.

```python
empleados = pd.DataFrame({
    "nombre": ["Ana","Luis","Carlos","Marta"],
    "departamento": ["IT","Ventas","IT","RRHH"],
    "cargo": ["Dev","Vendedor","Analista","HR"],
    "salario": [1000,1500,2000,1200],
    "experiencia": [2,6,10,4],
    "ciudad": ["Tegus","SPS","Tegus","SPS"]
})

print(empleados.head())

print("\nEstadísticas:")
print(empleados["salario"].describe())

print("\nPromedio por departamento:")
print(empleados.groupby("departamento")["salario"].mean())

print("\nMayor salario:")
print(empleados.loc[empleados["salario"].idxmax()])

print("\nExperiencia > 5:")
print(empleados[empleados["experiencia"] > 5])

empleados["Nivel"] = empleados["experiencia"].apply(
    lambda x: "Junior" if x < 3 else "Intermedio" if x <=5 else "Senior"
)

print("\nOrdenado por salario:")
print(empleados.sort_values(by="salario", ascending=False))
```
